# FiNTA — Data Exploration

Phân tích phân phối dữ liệu Bronze Layer trước khi labeling.

```python
import json
import pandas as pd

# Load labeled data (sau khi chạy llm_labeler.py)
with open("../data/labeled/checkpoint.json", encoding="utf-8") as f:
    data = json.load(f)

df = pd.DataFrame(data)
print(f"Tổng số bài đã gán nhãn: {len(df)}")
df.head()
```

```python
# Phân phối nhãn sentiment
print("Phân phối nhãn sentiment:")
print(df["sentiment_label"].value_counts())
print("\nTỷ lệ %:")
print(df["sentiment_label"].value_counts(normalize=True).round(3) * 100)
```

```python
# Phân phối confidence
print("Thống kê confidence:")
print(df["confidence"].describe())

threshold = 0.75
auto_accept = (df["confidence"] >= threshold).sum()
need_review = (df["confidence"] < threshold).sum()
print(f"\nAuto-accept (>= {threshold}): {auto_accept} ({auto_accept/len(df)*100:.1f}%)")
print(f"Cần review (< {threshold}): {need_review} ({need_review/len(df)*100:.1f}%)")
```

```python
# Phân phối theo nguồn tin (nếu có cột publisher)
if "publisher" in df.columns:
    print("Phân phối theo nguồn:")
    print(df["publisher"].value_counts())
```
